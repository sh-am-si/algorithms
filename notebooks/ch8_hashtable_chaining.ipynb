```python
class HashItem:
    def __init__(self, key, value):
        self.key = key
        self.value = value
```

```python
class Node:
    def __init__(self, key=None, value=None):
        self.key = key
        self.value = value
        self.next = None

    def __str__(self) -> str:
        return f"key={self.key} value={self.value}"
```

```python
class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def update(self, key, value) -> bool:
        """_summary_

        Args:
            key (_type_): _description_
            value (_type_): _description_

        Returns:
            bool: False added a new value
                  True updated existed 
        """
        node = Node(key, value)
        if self.head is None:
            self.head = node
            return False
        else:
            current = self.head
            while True:
                
                if current.key == key:
                    current.value = value
                    return True
                
                if current.next is None:
                    break
                else:
                    current = current.next
                
            current.next = node
            return False


    def traverse(self):
        current = self.head
        while current:
            print('"', current.key, "--", current.value, '"')
            current = current.next

    def search(self, key)-> Node | None:
        current = self.head
        while current:
            if current.key == key:
                print('"Record found:', current.key, "-", current.value, '"')
                return current
            current = current.next
        return None
```

```python
class HashTableChaining:
    def __init__(self):
        self.size = 6
        self.slots = [SinglyLinkedList() for i in range(self.size)]

    def _hash(self, key):
        mult = 1
        hv = 0
        for ch in key:
            hv += mult * ord(ch)
            mult += 1
        return hv % self.size

    def put(self, key, value):
        # self.resize()
        h = self._hash(key)
        self.slots[h].update(key, value)

    def get(self, key):
        h = self._hash(key)
        return self.slots[h].search(key)

    def printHashTable(self):
        print("Hash table is :- \n")
        print("Index \t\tValues\n")
        for x in range(self.size):
            print(x, end="\t\n")
            self.slots[x].traverse()

```

```python
ht = HashTableChaining()
ht.put("good", "eggs")
ht.put("better", "ham")
ht.put("best", "spam")
ht.put("ad", "do not")
ht.put("ga", "collide")
ht.put("awd", "do not")


v = ht.get("ad")
print(v)
```

```python
for key in ("good", "better", "best", "worst", "ad", "ga"):
    v = ht.get(key)
    print(v)


ht.printHashTable()
```
