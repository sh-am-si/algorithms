```python
class Stack:
    def __init__(self):
        self.elements = []

    def push(self, item):
        self.elements.append(item)

    def pop(self):
        return self.elements.pop()

    def __len__(self) -> int:
        return len(self.elements)


class TreeNode:
    def __init__(self, data=None):
        self.data = data
        self.right = None
        self.left = None

    def __repr__(self) -> str:
        return f"{'+' if self.left else '-'}:{self.data}:{'+' if self.right else '-'}"

    def __str__(self) -> str:
        return repr(self)


def calc(node):
    if node.data == "+":
        return calc(node.left) + calc(node.right)
    elif node.data == "-":
        return calc(node.left) - calc(node.right)
    elif node.data == "*":
        return calc(node.left) * calc(node.right)
    elif node.data == "/":
        return calc(node.left) // calc(node.right)  # INTEGER
    else:
        return node.data
```

```python
expr = "4 2 / 5 + 5 2 * 3 - *".split()

# (4 / 2 + 5) * ((5 * 2) - 3)

stack = Stack()

for term in expr:
    if term in "+-*/":
        node = TreeNode(term)
        node.right = stack.pop()
        node.left = stack.pop()
    else:
        node = TreeNode(int(term))
    stack.push(node)
```

```python
def inorder(root_node):
    current = root_node
    if current is None:
        return
    inorder(current.left)
    print(current)
    inorder(current.right)


inorder(stack.pop())
```
